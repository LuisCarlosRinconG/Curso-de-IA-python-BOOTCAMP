# Conceptos basicos de probabilidad

# Probabilidad condicional

La probabilidad condicional es un concepto muy relacionado con el Teorema de Bayes —de hecho, Bayes se basa en ella—.

Es La probabilidad de un evento A dado que el evento B ha ocurrido:

- P(A|B): Probabilidad de que ocurra A sabiendo que B ocurrió.
- P(A ∩ B): Probabilidad de que A y B ocurran al mismo tiempo.
- P(B): Probabilidad de que ocurra B.



![alt text](image-1.png)


# Teorema de Bayes

-P(A∣B) → Probabilidad de que ocurra A sabiendo que B ocurrió (probabilidad posterior).
- P(A): Probabilidad previa-Probabilidad de que A ocurra antes de ver la evidencia
- P(B|A): Verocimilitud-Probabilidad de observar B si A es cierto
- P(B): Evidencia-Probabilidad total de que ocurra B



![alt text](image.png)

```python
def bayes_theorem(pprevia,verocimilitud,evidencia):
    return (verocimilitud*pprevia)/evidencia

personas_enfermas=0.01
pruebas_positivas=0.99
falsos_positivos=0.05

print("el resultado es:",bayes_theorem(personas_enfermas,pruebas_positivas,falsos_positivos))
```
