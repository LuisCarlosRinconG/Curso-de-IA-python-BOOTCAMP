# Fundamentos de estadistica

# Tendencia central

La tendencia centra incluye:
- La media: que es el valor promedio de un conjunto de datos 
- La mediana: El valor medio cuando los datos estan ordenados
- La moda: El valor que ocurre con mas frecuencia

```python
data=[10,20,30,40,50,50,40,40]

mean=sum(data)/len(data)
print('La media es: ',mean)

sorted_data=sorted(data)
median=sorted_data[len(data)//2] if len(data)%2 !=0 else \
    (sorted_data[len(data)//2-1]+sorted_data[len(data)//2])/2
print('La mediana es: ',median)

#moda
from statistics import mode

print('La moda es: ', mode(data))
```

- Varianza: Es la desviación cuadrada proimedio de la media

En estadística, la varianza es una medida de dispersión que indica cuánto se alejan los datos de su media (promedio). Es decir, te dice qué tan “dispersos” o “concentrados” están los valores en un conjunto de datos.

🔹Interpretación

Si la varianza es baja, los datos están cerca de la media, hay poca variabilidad.

Si la varianza es alta, los datos están muy dispersos, hay mucha variabilidad.

🔹 ¿Para qué se utiliza?

La varianza se usa en muchos contextos:

Análisis estadístico básico: entender la variabilidad de los datos.

Finanzas: medir el riesgo de una inversión (cuánto varían los rendimientos).

Control de calidad: ver si un proceso es estable o presenta variaciones.

Machine Learning: en algoritmos como regresión, árboles o PCA (análisis de componentes principales).

Comparación de grupos: por ejemplo, en pruebas ANOVA (Análisis de Varianza).

```python
varianza=sum((x-mean)**2 for x in data) / len(data)

print('La varianza es: ', varianza)
```

- Desviación estandar: Es la raiz cuadrada de la varianza, indicando la disperción de los datos 

La desviación estándar es otra medida de dispersión, muy relacionada con la varianza, pero más intuitiva, porque está en las mismas unidades que los datos originales.

🔹 Definición

La desviación estándar mide cuánto se alejan, en promedio, los datos de la media.
Es la raíz cuadrada de la varianza

```python
std_dev=varianza**0.5
print('La desviación estandar es: ', std_dev)
```

# Pruebas de hipotesis (t-test)

Es un metodo estadistico pera determinar si un resultado es significativo o ocurrió por casualidad

- Pasos:

    - Formular la hipotesis: 
        - hipotesis nula: Asume que no se crea ningun efecto o diferenciua
        - hipotesis alternativa: La hipotesis alternativa asume que existe un efecto o diferencia
    - Calcular la estadistica de prueba que cuantifica la diferencia
    - Determinar P-Value: Probabilidad de observar datos tan extremos como las estadisticas de prueba bajo la hipotesis nula
    - Interpretación de resultados: se verifica si la probabilidad es menor que alfa(nivel de significancia(que sea mayor a 5% se rechaza el nivel de significancia))

## Intervalor de confianza y significancia estadistica
- Intervalo de confianza: Es un rango de valores dentro del cual se espera se encuentre el verdadero parametro de la población

```python
import scipy.stats as stats

sample_mean=mean
z_score=1.96 # nivel de confianza

ci=(sample_mean-z_score*(std_dev/len(data)**0.5),
    sample_mean + z_score*(std_dev/len(data) **0.5))

print("95 % de intervalo de confianza en nivel de IC",ci)
```

- Significancia estadistica: Un resultado es estadisticamente significativo si el valor P es menor que el nivel de significancia que es alfa igual a 0.05
