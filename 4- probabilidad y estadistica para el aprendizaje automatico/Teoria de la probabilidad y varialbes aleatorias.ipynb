# Conceptos basicos de probabilidad

# Teoria de la probabilidad

## Espacio muestral y eventos

- Espacio muestral: Se presenta como el conjunto de todos los resultados posibles de un experimento aleatorio. . Se denota generalmente como S o Ω.

- Eventos: Es un subconjunto del espacio muestral, es decir, un conjunto que contiene uno o varios resultados posibles del experimento.
Ejemplo: si el evento es “obtener un número par” al lanzar un dado, entonces 
A={2,4,6}.

## Probabilidad condicional

Es la probabilidad de que ocurra un evento A dado que B ha ocurrido. 


## Independiencia

la independencia se refiere a la ausencia de influencia entre dos o más eventos.
Es decir, que la ocurrencia de uno no afecta la probabilidad de que ocurra el otro.

P(A∩B)=P(A)×P(B)

Esto significa que la probabilidad de que ocurran ambos a la vez es igual al producto de sus probabilidades individuales.

```python
# Ejemplo de independencia

from itertools import product

# establecer un espacio mestral de un lanzamiento de dado
sample_space=list(range(1,7))

# probabilidad de sacar un ñumero par
n_pares=[2,4,6]

probabilidad_pares=len(n_pares)/len(sample_space)

print('La probabilidad de que caiga par',probabilidad_pares)
```

# Variables aleatorias

Una variable aleatoria asigna resultados de un experimento aleatorio a valores numericos

- Tipos: Discreto/ Continuo

Discreto: Toma un número contable de valores. Por ejemplo el ñumero de caras en 3 lanzamientos de moneda

Continuo: Toma valores de un rango continuo. Por ejemplo: La altura o el peso


- Función de masa de probabilidad(PMF):
Describe la distribución de probabilidad de una variable aleatoria discreta. Pj:

lanzar un dado donde la probabilidad es 1 de 6

- Función de densidad de probabilidad(PDF): Describe la distribución de probabilidad de una variable aleatoria continua que es el peso o la altura.

- Expectativa, varianza y desviación estandar:

-- Expectativa(E[X]): Es el promedio pomderado de los posibles valores de una variable aleatoria

resultados*probabilidades

-- Varianza(Var[X]): Donde x mide la dispersión de una variable aleatoria 

-- Desvoiación estandar: la raiz cuadrada de la varianza.

```python
import numpy as np

#rndom variable: lanzamiento de dado
resultados=np.array([1,2,3,4,5,6])
probabilidades=np.array([1/6]*6)

#Expectativa
expectativa=np.sum(resultados*probabilidades)
print('la expectativa es: ',expectativa)

#varianza y desviación estandar
varianza=np.sum((resultados-expectativa)**2*probabilidades)
desviación_estandar=np.sqrt(varianza)

print('La varianza es: ',varianza,' y la desviación estandar es: ',desviación_estandar)
```

# Ejercicio practico:

Calcular la probabilidad de que en 10.000 lanzamientos el tiro de dado me caiga:

- un número par
- Un número mayor que cuatro
-etc.

```python
import numpy as np

lanzamientos=np.random.randint(1,7,size=10000)

#calcular probabilidad:

par=np.sum(lanzamientos%2==0)/len(lanzamientos)

MayorACuatro=np.sum(lanzamientos>4)/len(lanzamientos)

print('la probabilidad de que caiga par es: ',par,' y la probabilidad ' \
'de que caiga un numero mayor a cuatro es: ',MayorACuatro)
print('El lanzamiento de dado fue: ',lanzamientos)
```
