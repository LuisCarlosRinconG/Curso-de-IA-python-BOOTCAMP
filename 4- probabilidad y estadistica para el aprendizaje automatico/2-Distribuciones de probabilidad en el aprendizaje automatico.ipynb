# Distribuciones de probabilidad comunes

## Distribución Gaussiana

La distribución gaussiana o distribución normal es una forma de representar cómo se distribuyen los datos alrededor de un valor promedio

Ejemplo:

Imagina medir la estatura de muchas personas:
La mayoría mide alrededor de 1.70 m (el promedio).
Algunas personas son un poco más altas o más bajas (por ejemplo, 1.80 o 1.60).
Muy pocas son extremadamente altas o muy bajas (como 2.00 o 1.40).
Si pones todas esas estaturas en un gráfico, verás una curva en forma de campana, con el centro en 1.70 m.
Esa es la distribución gaussiana

## Distribución binomial



La distribución binomial describe la probabilidad de tener cierto número de éxitos en una cantidad fija de intentos, cuando cada intento tiene solo dos posibles resultados:
✅ éxito o ❌ fracaso.


Lanzar una moneda 10 veces:
Cada lanzamiento tiene 2 resultados: cara o sello.
Puedes preguntar: “¿Cuál es la probabilidad de obtener exactamente 6 caras?”
👉 Ese tipo de pregunta se resuelve con la distribución binomial.


Características:

- Solo hay dos resultados posibles por intento (éxito o fracaso).
- Los intentos son independientes (uno no afecta al otro).
- La probabilidad de éxito (p) es la misma en todos los intentos.





## Distribución de Poisson

La distribución de Poisson se usa para modelar el número de veces que ocurre un evento en un intervalo fijo de tiempo, espacio o distancia,
cuando esos eventos suceden de manera aleatoria, independiente y con una tasa promedio constante.

Características:

- Los eventos son independientes entre sí.

- Ocurren con una tasa promedio constante (λ).

- No pueden ocurrir dos eventos exactamente al mismo tiempo (en teoría).

- El número de eventos posibles es entero y no negativo (0, 1, 2, …).


Ejemplos comunes:

📞 Número de llamadas que recibe un call center en una hora.

💥 Número de accidentes que ocurren en una carretera cada mes.

📧 Correos que te llegan en un día.

⚙️ Defectos en una línea de producción por cada lote.

En todos estos casos no sabes exactamente cuántos eventos ocurrirán,
pero sabes el promedio (por ejemplo, 10 llamadas por hora).










## Distribución uniforme

La distribución uniforme describe una situación en la que todos los valores posibles tienen la misma probabilidad de ocurrir.

En otras palabras, nada es más probable que otra cosa dentro de un rango determinado.









# Aplicación de distribuciones en machine learning

- Distribución gaussiana
Se utiliza en algoritmos como Naive Bayes y modelos de mezcla gaussiana

- Distribución bonomial
Es fundamental para la regresión logistica y otrros modelos de clasificación binaria

- Distribución de Poisson
Se aplica en el modelado de datos de conteo

- Distribución uniforme se utiliza en el muestreo aleatorio y la inicialización de parametros

# Visualización de distribuciones y comprensión de sus propiedades

Visualización ayuda a entender la asimetria, curtosis y valores atipicos

- Asimetria: Es la mediada de la simetria:
    - Asimetria positiva: Es la cola a la derecha
    - Asimetria negativa: Es la cola a la izquierda

- Kurtosis: Es la medida de la agudesa de la distribución:
    - Alta Kurtosis: Indica mas valores atipicos
    -Baja Kurtosis: Indica menos valores atipicos


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm,binom,poisson,uniform
import seaborn as sns

# Distribución Gaussiana

x=np.linspace(-4,4,100)
plt.plot(x,norm.pdf(x,loc=0,scale=1),label='Distribución gaussiana (u=0,s=1)')

# Distribución Binomial

n, p= 10,0.5
x=np.arange(0,n+1)
plt.bar(x,binom.pmf(x,n,p),alpha=0.7,label='Binomial (n=10,p=0.5)')

# Distribución de poisson
lam=3
x=np.arange(0,10)
plt.bar(x,poisson.pmf(x,lam),alpha=0.7,label='poisson(l=3)')

# Distribución uniforme (cambiar)
x=np.random.uniform(low=0,high=10,size=1000)
sns.histplot(x,kde=True,label='Uniforme',color='red')

plt.title('Distribuciones de probabilidad')
plt.legend()
plt.show()
```
