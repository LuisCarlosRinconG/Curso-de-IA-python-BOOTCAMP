# Distribuciones de probabilidad comunes

## Distribución Gaussiana

La distribución gaussiana o distribución normal es una forma de representar cómo se distribuyen los datos alrededor de un valor promedio

Ejemplo:

Imagina medir la estatura de muchas personas:
La mayoría mide alrededor de 1.70 m (el promedio).
Algunas personas son un poco más altas o más bajas (por ejemplo, 1.80 o 1.60).
Muy pocas son extremadamente altas o muy bajas (como 2.00 o 1.40).
Si pones todas esas estaturas en un gráfico, verás una curva en forma de campana, con el centro en 1.70 m.
Esa es la distribución gaussiana
